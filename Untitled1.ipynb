```python

# ==============================================================================
# PROJECT: NSE/BSE AI QUANTITATIVE TRADING & SENTIMENT AGENT (v2.0 - FULL ENGINE)
# ARCHITECT: Brijesh Kumar Patel | REGIME-AWARE MULTI-FACTOR QUANT SYSTEM
# ==============================================================================

!pip install -q yfinance pandas numpy plotly feedparser textblob

import yfinance as yf
import pandas as pd
import numpy as np
import feedparser
from textblob import TextBlob
import plotly.graph_objects as go
from plotly.subplots import make_subplots
from datetime import datetime
import warnings
warnings.filterwarnings('ignore')

print("🚀 Starting NSE/BSE Institutional Quant Agent v2.0...\n")

# ==============================================================================
# MODULE 1: PORTFOLIO & RISK CONFIGURATION
# ==============================================================================
CAPITAL = 100000          # Virtual Paper Trading Capital (₹1,00,000)
RISK_PER_TRADE = 0.02     # Max 2% Risk per trade (₹2,000)

WATCHLIST = [
    "RELIANCE.NS", "TCS.NS", "HDFCBANK.NS", "INFY.NS", "ICICIBANK.NS",
    "SBIN.NS", "BHARTIARTL.NS", "ITC.NS", "LT.NS", "TATAMOTORS.NS",
    "SUNPHARMA.NS", "BAJFINANCE.NS", "MARUTI.NS", "TITAN.NS", "WIPRO.NS",
    "ADANIENT.NS", "TATASTEEL.NS", "NTPC.NS", "POWERGRID.NS", "AXISBANK.NS"
]

# ==============================================================================
# MODULE 2: NIFTY 50 MARKET REGIME FILTER (MACRO BRAIN)
# ==============================================================================
def check_market_regime():
    nifty = yf.download("^NSEI", period="6mo", interval="1d", progress=False)
    if isinstance(nifty.columns, pd.MultiIndex):
        nifty.columns = nifty.columns.get_level_values(0)
    nifty['EMA_50'] = nifty['Close'].ewm(span=50, adjust=False).mean()
    curr_nifty = float(nifty['Close'].iloc[-1])
    ema50_nifty = float(nifty['EMA_50'].iloc[-1])

    if curr_nifty > ema50_nifty:
        return "🟢 BULLISH REGIME (Nifty > 50 EMA)", True, round(curr_nifty, 2)
    else:
        return "🔴 BEARISH / CAUTIOUS REGIME (Nifty < 50 EMA)", False, round(curr_nifty, 2)

regime_text, is_bullish_market, nifty_price = check_market_regime()
print(f"🏛️  NIFTY 50 LEVEL: {nifty_price} | MARKET REGIME: {regime_text}\n")

# ==============================================================================
# MODULE 3: TECHNICAL INDICATOR ENGINE
# ==============================================================================
def compute_indicators(df):
    df['EMA_20'] = df['Close'].ewm(span=20, adjust=False).mean()
    df['EMA_50'] = df['Close'].ewm(span=50, adjust=False).mean()

    # RSI (14)
    delta = df['Close'].diff()
    gain = (delta.where(delta > 0, 0)).rolling(window=14).mean()
    loss = (-delta.where(delta < 0, 0)).rolling(window=14).mean()
    rs = gain / loss
    df['RSI'] = 100 - (100 / (1 + rs))

    # MACD
    exp1 = df['Close'].ewm(span=12, adjust=False).mean()
    exp2 = df['Close'].ewm(span=26, adjust=False).mean()
    df['MACD'] = exp1 - exp2
    df['Signal_Line'] = df['MACD'].ewm(span=9, adjust=False).mean()

    # ATR (14)
    high_low = df['High'] - df['Low']
    high_close = np.abs(df['High'] - df['Close'].shift())
    low_close = np.abs(df['Low'] - df['Close'].shift())
    ranges = pd.concat([high_low, high_close, low_close], axis=1)
    df['ATR'] = np.max(ranges, axis=1).rolling(14).mean()

    # Volume Spike
    df['Vol_Avg'] = df['Volume'].rolling(20).mean()
    df['Vol_Ratio'] = df['Volume'] / df['Vol_Avg']
    return df

# ==============================================================================
# MODULE 4: NLP NEWS SENTIMENT ENGINE
# ==============================================================================
def get_news_sentiment(symbol):
    clean_name = symbol.replace(".NS", "")
    rss_url = f"https://google.com{clean_name}+share+news+india&hl=en-IN&gl=IN&ceid=IN:en"
    feed = feedparser.parse(rss_url)

    headlines = []
    scores = []
    for entry in feed.entries[:5]:
        headlines.append(entry.title)
        scores.append(TextBlob(entry.title).sentiment.polarity)

    avg_score = np.mean(scores) if scores else 0
    if avg_score > 0.05:
        label = "🟢 POSITIVE"
    elif avg_score < -0.05:
        label = "🔴 NEGATIVE"
    else:
        label = "🟡 NEUTRAL"

    top_news = headlines[0] if headlines else "No major news"
    return label, round(avg_score, 2), top_news

# ==============================================================================
# MODULE 5: MULTI-FACTOR SCANNER & RISK SIZER
# ==============================================================================
signals_list = []
top_stock_df = None
top_stock_name = ""
highest_score = -1

for ticker in WATCHLIST:
    try:
        df = yf.download(ticker, period="6mo", interval="1d", progress=False)
        if isinstance(df.columns, pd.MultiIndex):
            df.columns = df.columns.get_level_values(0)
        if len(df) < 55:
            continue

        df = compute_indicators(df)
        latest = df.iloc[-1]

        price = round(float(latest['Close']), 2)
        rsi = round(float(latest['RSI']), 2)
        ema20 = float(latest['EMA_20'])
        ema50 = float(latest['EMA_50'])
        macd = float(latest['MACD'])
        sig_line = float(latest['Signal_Line'])
        atr = float(latest['ATR'])
        vol_ratio = round(float(latest['Vol_Ratio']), 2)

        score = 0
        reasons = []

        if price > ema20 > ema50:
            score += 35
            reasons.append("Strong Uptrend")
        elif price > ema20:
            score += 20
            reasons.append("Above 20 EMA")

        if 42 <= rsi <= 65:
            score += 25
            reasons.append(f"Healthy RSI ({rsi})")
        elif rsi < 35:
            score += 30
            reasons.append(f"Oversold Bounce ({rsi})")

        if macd > sig_line:
            score += 25
            reasons.append("MACD Bullish")

        if vol_ratio > 1.1:
            score += 15
            reasons.append(f"Vol {vol_ratio}x")

        # Regime-Aware Signal Decision
        if score >= 70 and is_bullish_market:
            action = "🔥 STRONG BUY"
        elif score >= 70 and not is_bullish_market:
            action = "⚠️ BUY (HIGH RISK - BEAR MKT)"
        elif score >= 50:
            action = "✅ WATCHLIST"
        elif rsi > 75:
            action = "🔻 OVERBOUGHT"
        else:
            action = "⏳ NEUTRAL"

        if score >= 50 or rsi > 75:
            sent_label, sent_score, top_news = get_news_sentiment(ticker)
            if "NEGATIVE" in sent_label and "BUY" in action:
                action = "🚫 BLOCKED (NEG NEWS)"
        else:
            sent_label, sent_score, top_news = "⚪ SKIPPED", 0.0, "-"

        # ATR Risk Management (1:2 Risk-Reward)
        stop_loss = round(price - (1.5 * atr), 2)
        target = round(price + (3.0 * atr), 2)
        risk_per_share = max(price - stop_loss, 1)
        qty = int((CAPITAL * RISK_PER_TRADE) / risk_per_share)

        signals_list.append({
            "Stock": ticker.replace(".NS", ""),
            "Price (₹)": price,
            "Score": score,
            "Signal": action,
            "RSI": rsi,
            "Stop Loss (₹)": stop_loss,
            "Target (₹)": target,
            "Qty": qty,
            "News Sentiment": sent_label,
            "Setup Logic": ", ".join(reasons) if reasons else "Consolidating",
            "Top Headline": top_news[:60] + "..." if len(top_news) > 60 else top_news
        })

        if score > highest_score:
            highest_score = score
            top_stock_df = df.copy()
            top_stock_name = ticker.replace(".NS", "")

    except Exception:
        pass

results_df = pd.DataFrame(signals_list).sort_values(by="Score", ascending=False).reset_index(drop=True)

print("="*115)
print(f"📊 NSE/BSE AI QUANT AGENT v2.0 REPORT | {datetime.now().strftime('%d-%b-%Y %H:%M IST')}")
print("="*115)
display(results_df)

# ==============================================================================
# MODULE 6: INTERACTIVE INSTITUTIONAL CANDLESTICK CHART
# ==============================================================================
if top_stock_df is not None:
    print(f"\n📈 Generating Institutional Chart for Top Quant Pick: {top_stock_name} (Score: {highest_score}/100)...")
    plot_df = top_stock_df.tail(90)

    fig = make_subplots(rows=2, cols=1, shared_xaxes=True, vertical_spacing=0.05, row_heights=[0.7, 0.3],
                        subplot_titles=(f"{top_stock_name} - Institutional Price Action & EMAs", "RSI Momentum (14)"))

    fig.add_trace(go.Candlestick(x=plot_df.index, open=plot_df['Open'], high=plot_df['High'],
                                 low=plot_df['Low'], close=plot_df['Close'], name="Price"), row=1, col=1)
    fig.add_trace(go.Scatter(x=plot_df.index, y=plot_df['EMA_20'], line=dict(color='cyan', width=1.5), name="20 EMA"), row=1, col=1)
    fig.add_trace(go.Scatter(x=plot_df.index, y=plot_df['EMA_50'], line=dict(color='orange', width=1.5), name="50 EMA"), row=1, col=1)

    fig.add_trace(go.Scatter(x=plot_df.index, y=plot_df['RSI'], line=dict(color='violet', width=2), name="RSI"), row=2, col=1)
    fig.add_hline(y=70, line_dash="dash", line_color="red", row=2, col=1)
    fig.add_hline(y=30, line_dash="dash", line_color="green", row=2, col=1)

    fig.update_layout(height=620, template="plotly_dark", xaxis_rangeslider_visible=False,
                      title_text=f"AI Quant Agent v2.0 Visual Analysis: {top_stock_name}")
    fig.show()
```

```python

# ==============================================================================
# MODULE 7: 1-YEAR HISTORICAL BACKTESTING & PERFORMANCE ENGINE (v2.1)
# PURPOSE: Validates Strategy Win-Rate, Profit Factor & Net Return over 1 Year
# ==============================================================================

print("⏳ Running 1-Year Historical Backtest across NSE Watchlist...\n")

backtest_trades = []
INITIAL_CAPITAL = 100000
RISK_PER_TRADE = 0.02

for ticker in WATCHLIST:
    try:
        # Pichhle 1 saal ka daily data download karein
        df_bt = yf.download(ticker, period="1y", interval="1d", progress=False)
        if isinstance(df_bt.columns, pd.MultiIndex):
            df_bt.columns = df_bt.columns.get_level_values(0)
        if len(df_bt) < 60:
            continue

        df_bt = compute_indicators(df_bt)

        in_trade = False
        entry_price = 0
        stop_loss = 0
        target_price = 0
        entry_date = None
        qty = 0

        # Day-by-Day Simulation (Day 50 se lekar aaj tak)
        for i in range(50, len(df_bt)):
            row = df_bt.iloc[i]
            date_str = df_bt.index[i].strftime('%Y-%m-%d')

            # Agar trade chal raha hai toh check karein ki Target hit hua ya Stop-Loss
            if in_trade:
                if row['Low'] <= stop_loss:
                    pnl = round((stop_loss - entry_price) * qty, 2)
                    backtest_trades.append({
                        "Stock": ticker.replace(".NS", ""),
                        "Entry Date": entry_date,
                        "Exit Date": date_str,
                        "Entry (₹)": round(entry_price, 2),
                        "Exit (₹)": round(stop_loss, 2),
                        "Result": "❌ SL HIT",
                        "Net P&L (₹)": pnl
                    })
                    in_trade = False
                elif row['High'] >= target_price:
                    pnl = round((target_price - entry_price) * qty, 2)
                    backtest_trades.append({
                        "Stock": ticker.replace(".NS", ""),
                        "Entry Date": entry_date,
                        "Exit Date": date_str,
                        "Entry (₹)": round(entry_price, 2),
                        "Exit (₹)": round(target_price, 2),
                        "Result": "✅ TARGET HIT",
                        "Net P&L (₹)": pnl
                    })
                    in_trade = False

            # Naya Buy Signal check karein (Quant Score >= 70 logic)
            else:
                score = 0
                if row['Close'] > row['EMA_20'] > row['EMA_50']:
                    score += 35
                if 42 <= row['RSI'] <= 65:
                    score += 25
                if row['MACD'] > row['Signal_Line']:
                    score += 25
                if row['Vol_Ratio'] > 1.1:
                    score += 15

                if score >= 70 and pd.notna(row['ATR']):
                    in_trade = True
                    entry_price = float(row['Close'])
                    atr_val = float(row['ATR'])
                    stop_loss = entry_price - (1.5 * atr_val)
                    target_price = entry_price + (3.0 * atr_val)
                    risk_per_share = max(entry_price - stop_loss, 1)
                    qty = int((INITIAL_CAPITAL * RISK_PER_TRADE) / risk_per_share)
                    entry_date = date_str

    except Exception:
        pass

# Backtest Summary Report Taiyar Karein
bt_df = pd.DataFrame(backtest_trades)

if not bt_df.empty:
    total_trades = len(bt_df)
    wins = len(bt_df[bt_df['Net P&L (₹)'] > 0])
    losses = total_trades - wins
    win_rate = round((wins / total_trades) * 100, 2)
    total_pnl = round(bt_df['Net P&L (₹)'].sum(), 2)
    roi_pct = round((total_pnl / INITIAL_CAPITAL) * 100, 2)

    print("="*95)
    print("🏆 1-YEAR HISTORICAL BACKTEST SCORECARD (PAPER CAPITAL: ₹1,00,000)")
    print("="*95)
    print(f"📌 Total Completed Trades : {total_trades}")
    print(f"✅ Winning Trades         : {wins}")
    print(f"❌ Losing Trades          : {losses}")
    print(f"🎯 Strategy Win-Rate      : {win_rate}% (1:2 Risk-to-Reward)")
    print(f"💰 Net Simulated Profit   : ₹{total_pnl:,.2f} ({roi_pct}% Return)")
    print("="*95)
    print("\n📋 Recent 15 Simulated Trades Log:")
    display(bt_df.tail(15).reset_index(drop=True))
else:
    print("⚠️ Is period me koi completed trade nahi mila.")
```

```python

# ==============================================================================
# MODULE 8: INTRADAY 15-MINUTE MOMENTUM & VWAP SCANNER (v2.2)
# PURPOSE: Generates Fast Intraday Setups with Tighter 1:1.5 Risk-Reward
# ==============================================================================

print("⚡ Running Intraday 15-Minute Scanner across Liquid NSE Stocks...\n")

intraday_signals = []

for ticker in WATCHLIST:
    try:
        # Pichhle 5 din ka 15-minute intraday data download karein
        df_intra = yf.download(ticker, period="5d", interval="15m", progress=False)
        if isinstance(df_intra.columns, pd.MultiIndex):
            df_intra.columns = df_intra.columns.get_level_values(0)
        if len(df_intra) < 30:
            continue

        df_intra = compute_indicators(df_intra)

        # Intraday VWAP (Volume Weighted Average Price) Calculation
        df_intra['Cum_Vol'] = df_intra['Volume'].cumsum()
        df_intra['Cum_Vol_Price'] = (df_intra['Close'] * df_intra['Volume']).cumsum()
        df_intra['VWAP'] = df_intra['Cum_Vol_Price'] / df_intra['Cum_Vol']

        last = df_intra.iloc[-1]
        price = round(float(last['Close']), 2)
        vwap = round(float(last['VWAP']), 2)
        rsi = round(float(last['RSI']), 2)
        ema20 = float(last['EMA_20'])
        macd = float(last['MACD'])
        sig = float(last['Signal_Line'])
        atr = float(last['ATR'])

        intra_score = 0
        setup_notes = []

        if price > vwap:
            intra_score += 35
            setup_notes.append("Above VWAP")
        if price > ema20:
            intra_score += 25
            setup_notes.append("Above 20 EMA")
        if 45 <= rsi <= 68:
            intra_score += 20
            setup_notes.append(f"Momentum RSI ({rsi})")
        if macd > sig:
            intra_score += 20
            setup_notes.append("MACD Cross")

        if intra_score >= 80:
            signal = "🚀 INTRADAY BUY"
        elif intra_score <= 25 and price < vwap:
            signal = "🔻 INTRADAY SHORT"
        else:
            signal = "⏳ WAIT"

        # Tight Intraday Stop-Loss (1x ATR) & Target (1.5x ATR)
        sl = round(price - (1.0 * atr), 2) if "BUY" in signal else round(price + (1.0 * atr), 2)
        tgt = round(price + (1.5 * atr), 2) if "BUY" in signal else round(price - (1.5 * atr), 2)

        intraday_signals.append({
            "Stock": ticker.replace(".NS", ""),
            "LTP (₹)": price,
            "VWAP (₹)": vwap,
            "Intra Score": intra_score,
            "Intraday Signal": signal,
            "RSI (15m)": rsi,
            "Tight SL (₹)": sl,
            "Target (₹)": tgt,
            "Setup": " + ".join(setup_notes) if setup_notes else "Below VWAP / Weak"
        })
    except Exception:
        pass

intra_df = pd.DataFrame(intraday_signals).sort_values(by="Intra Score", ascending=False).reset_index(drop=True)

print("="*105)
print(f"⚡ NSE INTRADAY (15-MIN) LIVE SCANNER | {datetime.now().strftime('%d-%b-%Y %H:%M IST')}")
print("="*105)
display(intra_df)
```

```python

# ==============================================================================
# MODULE 9: FULL NIFTY 50 EXPANSION & AUTO-CSV REPORT EXPORTER (FINAL v3.0)
# PURPOSE: Scans all 50 Nifty Constituents & Saves Daily Signal Sheet
# ==============================================================================

NIFTY_50_WATCHLIST = [
    "RELIANCE.NS", "TCS.NS", "HDFCBANK.NS", "ICICIBANK.NS", "BHARTIARTL.NS",
    "SBIN.NS", "INFY.NS", "LICI.NS", "ITC.NS", "HINDUNILVR.NS",
    "LT.NS", "BAJFINANCE.NS", "HCLTECH.NS", "MARUTI.NS", "SUNPHARMA.NS",
    "ADANIENT.NS", "KOTAKBANK.NS", "TITAN.NS", "ONGC.NS", "TATAMOTORS.NS",
    "NTPC.NS", "AXISBANK.NS", "ULTRACEMCO.NS", "ADANIPORTS.NS", "POWERGRID.NS",
    "M&M.NS", "BAJAJFINSV.NS", "WIPRO.NS", "COALINDIA.NS", "NESTLEIND.NS",
    "JSWSTEEL.NS", "TATASTEEL.NS", "SIEMENS.NS", "IRFC.NS", "VBL.NS",
    "TRENT.NS", "ZOMATO.NS", "HAL.NS", "BEL.NS", "DLF.NS",
    "GRASIM.NS", "TECHM.NS", "HINDALCO.NS", "ASIANPAINT.NS", "SBILIFE.NS",
    "BAJAJ-AUTO.NS", "EICHERMOT.NS", "DIVISLAB.NS", "DRREDDY.NS", "CIPLA.NS"
]

print(f"🔎 Scanning Complete Nifty 50 Universe ({len(NIFTY_50_WATCHLIST)} Stocks)...")

final_report = []

for ticker in NIFTY_50_WATCHLIST:
    try:
        df_n50 = yf.download(ticker, period="6mo", interval="1d", progress=False)
        if isinstance(df_n50.columns, pd.MultiIndex):
            df_n50.columns = df_n50.columns.get_level_values(0)
        if len(df_n50) < 55:
            continue

        df_n50 = compute_indicators(df_n50)
        last = df_n50.iloc[-1]

        price = round(float(last['Close']), 2)
        rsi = round(float(last['RSI']), 2)
        ema20 = float(last['EMA_20'])
        ema50 = float(last['EMA_50'])
        macd = float(last['MACD'])
        sig = float(last['Signal_Line'])
        atr = float(last['ATR'])
        vol_ratio = round(float(last['Vol_Ratio']), 2)

        score = 0
        if price > ema20 > ema50: score += 35
        elif price > ema20: score += 20
        if 42 <= rsi <= 65: score += 25
        elif rsi < 35: score += 30
        if macd > sig: score += 25
        if vol_ratio > 1.1: score += 15

        if score >= 70:
            action = "🔥 STRONG BUY"
            sent_label, _, top_news = get_news_sentiment(ticker)
            if "NEGATIVE" in sent_label:
                action = "🚫 BLOCKED (NEG NEWS)"
        elif score >= 50:
            action = "✅ WATCHLIST"
            sent_label, top_news = "⚪ NOT CHECKED", "-"
        else:
            action = "⏳ NEUTRAL"
            sent_label, top_news = "⚪ SKIPPED", "-"

        stop_loss = round(price - (1.5 * atr), 2)
        target = round(price + (3.0 * atr), 2)
        qty = int((CAPITAL * RISK_PER_TRADE) / max(price - stop_loss, 1))

        final_report.append({
            "Date": datetime.now().strftime('%Y-%m-%d'),
            "Stock": ticker.replace(".NS", ""),
            "Close (₹)": price,
            "Quant Score": score,
            "Action": action,
            "RSI": rsi,
            "Stop Loss (₹)": stop_loss,
            "Target (₹)": target,
            "Qty": qty,
            "News Sentiment": sent_label
        })
    except Exception:
        pass

final_df = pd.DataFrame(final_report).sort_values(by="Quant Score", ascending=False).reset_index(drop=True)

# CSV File Save Karein (Paper Trading Journal ke liye)
file_name = f"NSE_Quant_Signals_{datetime.now().strftime('%Y_%m_%d')}.csv"
final_df.to_csv(file_name, index=False)

print(f"\n✅ Scan Complete! Report saved as: {file_name}")
print("🏆 TOP 10 HIGH-CONVICTION STOCKS TODAY:")
display(final_df.head(10))
```
